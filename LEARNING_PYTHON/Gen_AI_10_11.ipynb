```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.linear_model import LinearRegression
data = {
    "study-hour" : [1, 2, 3, 4,5,6, 7,8, 9],
    "marks" : [35, 45, 52, 64, 73, 80, 88, 92, 98]
}
df= pd.DataFrame(data)
df


plt.plot(df["study-hour"], df["marks"], color="r", linestyle="--")

sns.scatterplot(data=df, x="study-hour", y="marks")

plt.xlabel("Study Hours -->")
plt.ylabel("Marks -->")

plt.show()
```

```python
X= df[["study-hour"]]
y=  df["marks"]

```

```python
model = LinearRegression()
model
```

```python
#model  trained
model.fit(X, y) #fit is used fro trainig purpose
# if X and y are categorical then must be convert into numerical is caleed feature Enginerring
```

```python
c = model.intercept_
print(round(c, 2))
```

```python
#y = m+c
# m is ceofiicent of x

m = model.coef_[0]
print(round(m,2))
```

```python
# formula apply y = mx +c
study_hour = float(input("Enter a sutdyhours"))
y = m*study_hour +c
print("Predicated Value is ", round(y, 2))
```

# **In Linear Regression, the 3 common types of error (residual/error) are:**

MAE – Mean Absolute Error

Measures the average absolute difference between actual and predicted values.
Formula:

# $$ MAE = \frac{1}{n}\sum |y-\hat{y}| $$

MSE – Mean Squared Error

Squares each error, so larger errors are penalized more.
Formula:
# **$$ MSE = \frac{1}{n}\sum (y-\hat{y})^2 $$**

RMSE – Root Mean Squared Error

Square root of MSE.
Its unit is the same as the target variable.
Formula:

# $$ RMSE = \sqrt{\frac{1}{n}\sum (y-\hat{y})^2} $$

```python
MAE = (compare_df["Actual Y"] - compare_df["Predicted Y"]).abs().mean()
print("Mean Absolute Error", round(MAE, 2))
```

```python
MSE = ((compare_df["Actual Y"] - compare_df["Predicted Y"])**2).abs().mean()
print("Mean Squared Error", round(MSE, 2))
```

```python
import math
RMSE = math.sqrt(MSE)
print("Root Mean Squared Error", round(RMSE, 2))
```
