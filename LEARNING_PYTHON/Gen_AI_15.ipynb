# **Last technique for data preprocessing using mean and median**

# ***9th Interview :-***


Why mean and median used for data preprocessing


and :- mean and median always working with numerical data

mean  and median always working in data preprocessing for NaN value not for 0

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "age" : [20, 21, np.nan, 22, 21],
    "marks" : [80, 90, np.nan , 75, np.nan]
}
df = pd.DataFrame(data)
df
```

```python
df["age"] = df["age"].mean()
df
```

```python
df["marks"] = df["marks"].median()
df
```

```python
df["marks"] = df["marks"].fillna(df["marks"].median())
df
```
