# Inheritance : -- The process of acquiring properties and behaviour
# **from one to another class is Known as inheritance.**

# **There are five tyoes of inheritance:**

# 1.single level inheritance:- when a class inherits the properities of another

# **2.multilevel- level inheritance: - when a class inherites the properities of another but these class already inherited another class**

# **3.Multiple inheritance:- when a class inherits more than one class.**

# **4.Hierarcical Inheritance :-  when  a class inherited by more than one class**

# **5.Hybrid Inheritance:- it is a combination of hierachical anf multiple inheritance.**

```python
class A:
  def m1(self):
    print("hello")
  def m2(self, a, b):
    print(self.a + self.b)
class B(A):
  pass
b = B()
b.m1()
b.m2()
```

```python
class A:
  def show(self):
    print("Welcome")
  def multiply(self, a, b):
    print(self.a*self.b)
class B(A):
  def add(self):
    print("bye")
class C(B):
  pass
c = C()
c.show()
c.add()
```

```python
class Student:
  def __init__(self, name, course):
    self.name = name
    self.course = course
  class Address:
    def __init__(self, city, state):
      self.city = city
      self.state = state
    def show(self):
      print("City is", self.city)
      print("State is", self.state)
s = Student("rahul", "java")
a = s.Address("pune", "maharashtra") # Instantiate Address via a Student object
a.show()
print("Name is",s.name)
print("Course is",s.course)
```

```python
class ShoppingCart:
  def __init__(self): # This is the __init__ for ShoppingCart
    self.cart = [] # Initialize cart here

  class Product:
    def __init__(self, name, price, quantity): # This is the correct __init__ for Product
      self.name = name
      self.price = price
      self.quantity = quantity
    def total(self):
      return self.price * self.quantity

  def add_product(self): # Correctly indented as a method of ShoppingCart
    name = input("Enter product name: ")
    price = int(input("Enter product price: "))
    quantity = int(input("Enter product quantity: "))

    p = self.Product(name, price, quantity)
    self.cart.append(p)
    print("Product Added Successfully")

  def show_cart(self): # Correctly indented as a method of ShoppingCart
    for p in self.cart:
      print(p.name, p.price, p.quantity, p.total())


shop = ShoppingCart()
shop.add_product()
shop.show_cart()
```
