## **Real Time Application Using class and object**

```python
class Customer:
  bname = "HDFC Bank Mohali"
  def __init__(self, name, balance = 0):
    self.name = name
    self.balance = balance
  def deposit(self, amt):
    self.balance = self.balance+amt
    print("After Deposit Balance is", self.balance)
  def withdraw(self, amt):
    if self.balance < amt:
      print("Insufficent Balance")
    self.balance=self.balance-amt
    print("After Withdraw Balance is", self.balance)
name = input("Enter Customer name")
c = Customer(name)
print("Welcome to", Customer.bname,"Mr", name)
while True:
  print("d- deposit")
  print("w-Withdraw")
  print("e-exit")
  ch = input("Enter your choice")
  if ch == 'd' or ch == 'D':
    amt = int(input("Enter amount"))
    c.deposit(amt)
  elif ch == 'w' or ch == 'W':
    amt = int(input("Enter amount for withraw"))
    c.withdraw(amt)
  elif ch =='e' or ch == 'E':
    print("Thank for using HDFC ATM Mohali")
    break
  else:
    print("Invalid choice")



```

```python
class Student:
    school = "ABC College Mohali"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)

    def grade(self):
        if self.marks >= 90:
            print("Grade: A")
        elif self.marks >= 75:
            print("Grade: B")
        elif self.marks >= 60:
            print("Grade: C")
        elif self.marks >= 50:
            print("Grade: D")
        else:
            print("Grade: F")


name = input("Enter Student Name: ")
marks = int(input("Enter Student Marks: "))

s = Student(name, marks)

print("Welcome to", Student.school)

while True:
    print("\n1 - Display Student")
    print("2 - Check Grade")
    print("3 - Exit")

    ch = input("Enter your choice: ")

    if ch == '1':
        s.display()

    elif ch == '2':
        s.grade()

    elif ch == '3':
        print("Thank you for using Student Grade System")
        break

    else:
        print("Invalid choice")
```
