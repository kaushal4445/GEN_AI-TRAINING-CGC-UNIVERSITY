```python
from abc import ABC, abstractmethod
class Demo(ABC):
  @abstractmethod  # predefined like save(), fetchALL(), findByID
  def show(self):
    pass
  @abstractmethod
  def msg(self):
    pass
class Test(Demo):
  def show(self):
    print("this is show method")
  def msg(self):
    print("this is msg method")
t=Test()
t.show()
t.msg()
```

```python
class Test1(ABC):
  def __init__(self):
    self.a = 10
    self.b =20
  @abstractmethod
  def show():
    pass
class Test2(Test1):
  def show(self):
    print("Value of A:", self.a, "Value of  B:", self.b)

t2 = Test2()
t2.show()
```

```python
from abc import ABC, abstractmethod
class Test3(ABC):
  @abstractmethod
  def m1():
    pass
class Test4(ABC):
  @abstractmethod
  def m2():
    pass
class Test5(Test3, Test4):
  def m1(self):
    print("This id 3rd class")
  def m2(self):
    print("This id 4th class")
t5 = Test5()
t5.m1()
t5.m2()

```

```python
def vote():
  try:
    age = int(input("Enter age :"))
    if age < 18:
      raise ValueError
    else:
        print("You are eligible to vote")
  except ValueError:
    print("You are not eligible to vote")
vote()

```

```python
def applylicencise():
  try:
    age = int(input("Enter age :"))
    documents = input("Enter documents is having or not :")

    if age < 18:
      raise ValueError
    else:
      print("You are eligible to apply for license (age-wise).")

    if documents.lower() == 'no' or documents.upper() == 'NO':
      raise Exception
    else:
      print("You are eligible (documents-wise).")

  except ValueError:
    print("You are not eligible for license (age-related).")
  except Exception:
    print("You are not having documents (document-related).")

applylicencise()
```
